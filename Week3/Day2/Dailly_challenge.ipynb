Daily Challenge: Data Handling and Analysis in Python

Your Task

Download and import the Data Science Job Salary dataset.
Normalize the ‘salary’ column using Min-Max normalization which scales all salary values between 0 and 1.
Implement dimensionality reduction like Principal Component Analysis (PCA) or t-SNE to reduce the number of features (columns) in the dataset.
Group the dataset by the ‘experience_level’ column and calculate the average and median salary for each experience level (e.g., Junior, Mid-level, Senior).


```python
import pandas as pd
```

```python
salary_df=pd.read_csv("Datasets/datascience_salaries.csv")

salary_df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

salary_df["salary_normalized"] = scaler.fit_transform(
    salary_df[["salary"]]
).ravel()

salary_df[["salary", "salary_normalized"]].describe()
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

numeric_df = salary_df.select_dtypes(include="number").drop(
    columns=["salary_normalized"],
    errors="ignore"
)

numeric_df = numeric_df.fillna(numeric_df.median())

scaled_data = StandardScaler().fit_transform(numeric_df)

pca = PCA(n_components=2)

pca_result = pca.fit_transform(scaled_data)

pca_df = pd.DataFrame(
    pca_result,
    columns=["PC1", "PC2"],
    index=salary_df.index
)

pca_df.head()
```

```python
display("Original features:", numeric_df.shape[1])
display("PCA components:", pca_df.shape[1])
display("Explained variance:", pca.explained_variance_ratio_.sum())
```

```python
salary_summary = (
    salary_df.groupby("experience_level", as_index=False)
    .agg(
        Average_Salary=("salary", "mean"),
        Median_Salary=("salary", "median")
    )
)

salary_summary
```
