### Challenge Description:

Scenario: You are a data analyst working with a dataset of global weather. Your task is to analyze temperature trends and visualize the results.

#### 1. Data Preparation:

Hint 1: Use np.random.uniform(low, high, size) to generate the temperature data.
Hint 2: Create a DataFrame using pd.DataFrame(data, index, columns) with appropriate index and columns.

Use NumPy to generate a synthetic dataset representing average monthly temperatures (in degrees Celsius) for 12 months across 10 different cities. The temperatures should range from -5 to 35 degrees.

Convert this NumPy array into a Pandas DataFrame, adding city names as index and months as columns.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
```

```python
temp=rng.uniform(-5,35,size=(10,12))
display(temp)
```

```python
import calendar

months= list(calendar.month_name)[1:]

df=pd.DataFrame(temp,
                index=('Tel Aviv','Haifa','Jerusalem','Eilat','Beer Sheva','Akko','Zfat','Bat Yam','Dimona','Ramat Gan'),
                       columns=months
                       )
display(df)
```

#### 2. Data Analysis:

Hint 1: Calculate the annual average temperature using DataFrame.mean(axis).
Hint 2: Find the city with the highest and lowest average temperature using idxmax() and idxmin() methods.

Calculate the annual average temperature for each city.

Identify the city with the highest and lowest average temperature for the year.

```python
df["Average"]=df[months].mean(axis=1)

display(df)
```

```python
warmest_city = df["Average"].idxmax()
coldest_city = df["Average"].idxmin()

print("Warmest city:", warmest_city)
print("Coldest city:", coldest_city)
```

#### 3. Data Visualization:

```python
plt.figure(figsize=(12, 6))

df["Average"].sort_values().plot(
    kind="barh",
    color="orange",
    edgecolor="black"
)

plt.title("Average Annual Temperature by City")
plt.xlabel("Average Temperature (°C)")
plt.ylabel("City")
plt.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 7))

for city in df.index:
    plt.plot(
        months,
        df.loc[city, months],
        marker="o",
        label=city
    )

plt.title("Monthly Temperature Trends Across Cities")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.xticks(rotation=45)
plt.legend(title="Cities", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Conclusion:

This project analyzed synthetic monthly temperature data for 10 cities using NumPy, Pandas, and Matplotlib.
Dimona had the highest annual average temperature (approximately 18.8°C), while Tel Aviv had the lowest (approximately 13.7°C). The monthly temperature chart showed significant fluctuations across all cities.
No consistent seasonal pattern was observed, as the temperatures were randomly generated using a uniform distribution.
